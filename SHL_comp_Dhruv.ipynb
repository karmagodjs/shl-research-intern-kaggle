```python
import os
import pandas as pd
from tqdm.notebook import tqdm

if os.path.exists(TRAIN_TRANSCRIPT_CSV):
    train_transcripts_df = pd.read_csv(TRAIN_TRANSCRIPT_CSV)
    processed_files = set(train_transcripts_df["filename"])
else:
    train_transcripts_df = pd.DataFrame(
        columns=["filename", "label", "transcript"]
    )
    processed_files = set()

for _, row in tqdm(
    train_df.iterrows(),
    total=len(train_df),
    desc="Transcribing training audio"
):
    filename = row["filename"]

    if filename in processed_files:
        continue

    audio_path = os.path.join(TRAIN_AUDIO_DIR, filename)

    try:
        result = whisper_model.transcribe(
            audio_path,
            fp16=False
        )

        new_row = pd.DataFrame([{
            "filename": filename,
            "label": row["label"],
            "transcript": result["text"].strip()
        }])

        train_transcripts_df = pd.concat(
            [train_transcripts_df, new_row],
            ignore_index=True
        )

        train_transcripts_df.to_csv(
            TRAIN_TRANSCRIPT_CSV,
            index=False
        )

    except Exception as e:
        print(f"\nError processing {filename}: {e}")

print("\nTraining transcription completed.")
print("Number of transcripts:", len(train_transcripts_df))
```

```python
import re
import numpy as np
import pandas as pd

FILLER_WORDS = {
    "um", "uh", "erm", "hmm", "like",
    "you know", "actually", "basically"
}

SELF_CORRECTION_PHRASES = {
    "i mean",
    "or rather",
    "sorry",
    "what i mean",
    "let me rephrase",
    "i mean to say"
}


def extract_advanced_linguistic_features(text):
    text = str(text).strip()
    doc = nlp(text)

    features = {}

    words = [
        token.text.lower()
        for token in doc
        if token.is_alpha
    ]

    word_count = len(words)
    sentences = list(doc.sents)
    sentence_count = len(sentences)

    # --------------------------------------------------
    # Basic statistics
    # --------------------------------------------------

    features["word_count"] = word_count
    features["unique_word_count"] = len(set(words))

    features["type_token_ratio"] = (
        len(set(words)) / word_count
        if word_count else 0
    )

    features["sentence_count"] = sentence_count

    sentence_lengths = []

    for sent in sentences:
        sent_words = [
            token for token in sent
            if token.is_alpha
        ]
        sentence_lengths.append(len(sent_words))

    features["avg_sentence_length"] = (
        np.mean(sentence_lengths)
        if sentence_lengths else 0
    )

    features["sentence_length_std"] = (
        np.std(sentence_lengths)
        if sentence_lengths else 0
    )

    features["max_sentence_length"] = (
        max(sentence_lengths)
        if sentence_lengths else 0
    )

    features["min_sentence_length"] = (
        min(sentence_lengths)
        if sentence_lengths else 0
    )

    # --------------------------------------------------
    # POS distribution
    # --------------------------------------------------

    pos_counts = {}

    for token in doc:
        if token.is_alpha:
            pos_counts[token.pos_] = (
                pos_counts.get(token.pos_, 0) + 1
            )

    for pos in [
        "NOUN", "VERB", "AUX", "ADJ",
        "ADV", "PRON", "DET", "ADP",
        "CCONJ", "SCONJ", "NUM"
    ]:

        count = pos_counts.get(pos, 0)

        features[f"{pos.lower()}_count"] = count

        features[f"{pos.lower()}_ratio"] = (
            count / word_count
            if word_count else 0
        )

    # --------------------------------------------------
    # Verb complexity
    # --------------------------------------------------

    verbs = [
        token for token in doc
        if token.pos_ == "VERB"
    ]

    finite_verbs = [
        token for token in doc
        if token.pos_ in {"VERB", "AUX"}
        and token.morph.get("VerbForm")
    ]

    features["verb_count"] = len(verbs)

    features["finite_verb_count"] = len(finite_verbs)

    features["verbs_per_sentence"] = (
        len(verbs) / sentence_count
        if sentence_count else 0
    )

    # --------------------------------------------------
    # Clause / sentence complexity
    # --------------------------------------------------

    subordinate_count = sum(
        1 for token in doc
        if token.dep_ in {
            "advcl",
            "ccomp",
            "xcomp",
            "acl",
            "relcl"
        }
    )

    features["subordinate_clause_count"] = subordinate_count

    features["subordinate_clause_ratio"] = (
        subordinate_count / sentence_count
        if sentence_count else 0
    )

    # --------------------------------------------------
    # Dependency structure
    # --------------------------------------------------

    dependency_counts = {}

    for token in doc:
        dependency_counts[token.dep_] = (
            dependency_counts.get(token.dep_, 0) + 1
        )

    features["root_count"] = dependency_counts.get(
        "ROOT", 0
    )

    features["subject_count"] = (
        dependency_counts.get("nsubj", 0)
        + dependency_counts.get("nsubjpass", 0)
    )

    features["object_count"] = (
        dependency_counts.get("dobj", 0)
        + dependency_counts.get("obj", 0)
    )

    # --------------------------------------------------
    # Sentence completeness proxies
    # --------------------------------------------------

    complete_sentence_count = 0

    for sent in sentences:

        has_subject = any(
            token.dep_ in {"nsubj", "nsubjpass", "expl"}
            for token in sent
        )

        has_predicate = any(
            token.pos_ in {"VERB", "AUX"}
            for token in sent
        )

        if has_subject and has_predicate:
            complete_sentence_count += 1

    features["complete_sentence_ratio"] = (
        complete_sentence_count / sentence_count
        if sentence_count else 0
    )

    features["incomplete_sentence_ratio"] = (
        1 - features["complete_sentence_ratio"]
        if sentence_count else 0
    )

    # --------------------------------------------------
    # Repetition
    # --------------------------------------------------

    adjacent_repetition = sum(
        1 for i in range(1, len(words))
        if words[i] == words[i - 1]
    )

    features["adjacent_repetition"] = adjacent_repetition

    features["repetition_ratio"] = (
        adjacent_repetition / word_count
        if word_count else 0
    )

    # --------------------------------------------------
    # Fillers
    # --------------------------------------------------

    filler_words = {
        "um", "uh", "erm", "hmm",
        "like", "actually", "basically"
    }

    filler_count = sum(
        words.count(word)
        for word in filler_words
    )

    features["filler_count"] = filler_count

    features["filler_ratio"] = (
        filler_count / word_count
        if word_count else 0
    )

    # --------------------------------------------------
    # Self-correction
    # --------------------------------------------------

    text_lower = text.lower()

    correction_phrases = [
        "i mean",
        "or rather",
        "sorry",
        "what i mean",
        "let me rephrase"
    ]

    correction_count = sum(
        text_lower.count(phrase)
        for phrase in correction_phrases
    )

    features["self_correction_count"] = correction_count

    # --------------------------------------------------
    # Function-word usage
    # --------------------------------------------------

    function_words = {
        "the", "a", "an",
        "is", "am", "are",
        "was", "were",
        "to", "of", "in",
        "on", "for", "with"
    }

    function_word_count = sum(
        word in function_words
        for word in words
    )

    features["function_word_ratio"] = (
        function_word_count / word_count
        if word_count else 0
    )

    # --------------------------------------------------
    # Sentence length categories
    # --------------------------------------------------

    if sentence_lengths:

        features["short_sentence_ratio"] = (
            np.mean(np.array(sentence_lengths) < 5)
        )

        features["medium_sentence_ratio"] = (
            np.mean(
                (np.array(sentence_lengths) >= 5) &
                (np.array(sentence_lengths) <= 20)
            )
        )

        features["long_sentence_ratio"] = (
            np.mean(np.array(sentence_lengths) > 20)
        )

    else:

        features["short_sentence_ratio"] = 0
        features["medium_sentence_ratio"] = 0
        features["long_sentence_ratio"] = 0

    return features
```

```python
import librosa
import numpy as np
import pandas as pd


def extract_acoustic_features(audio_path, sr=16000):

    y, _ = librosa.load(
        audio_path,
        sr=sr,
        mono=True
    )

    features = {}

    # --------------------------------------------------
    # Duration
    # --------------------------------------------------

    duration = len(y) / sr

    features["audio_duration"] = duration

    # --------------------------------------------------
    # RMS energy
    # --------------------------------------------------

    rms = librosa.feature.rms(y=y)[0]

    features["rms_mean"] = np.mean(rms)
    features["rms_std"] = np.std(rms)
    features["rms_min"] = np.min(rms)
    features["rms_max"] = np.max(rms)

    # --------------------------------------------------
    # Zero crossing rate
    # --------------------------------------------------

    zcr = librosa.feature.zero_crossing_rate(y)[0]

    features["zcr_mean"] = np.mean(zcr)
    features["zcr_std"] = np.std(zcr)

    # --------------------------------------------------
    # Spectral features
    # --------------------------------------------------

    spectral_centroid = librosa.feature.spectral_centroid(
        y=y,
        sr=sr
    )[0]

    spectral_bandwidth = librosa.feature.spectral_bandwidth(
        y=y,
        sr=sr
    )[0]

    spectral_rolloff = librosa.feature.spectral_rolloff(
        y=y,
        sr=sr
    )[0]

    features["spectral_centroid_mean"] = np.mean(
        spectral_centroid
    )

    features["spectral_centroid_std"] = np.std(
        spectral_centroid
    )

    features["spectral_bandwidth_mean"] = np.mean(
        spectral_bandwidth
    )

    features["spectral_bandwidth_std"] = np.std(
        spectral_bandwidth
    )

    features["spectral_rolloff_mean"] = np.mean(
        spectral_rolloff
    )

    features["spectral_rolloff_std"] = np.std(
        spectral_rolloff
    )

    # --------------------------------------------------
    # MFCC
    # --------------------------------------------------

    mfcc = librosa.feature.mfcc(
        y=y,
        sr=sr,
        n_mfcc=13
    )

    for i in range(13):

        features[f"mfcc_{i+1}_mean"] = np.mean(
            mfcc[i]
        )

        features[f"mfcc_{i+1}_std"] = np.std(
            mfcc[i]
        )

    # --------------------------------------------------
    # MFCC delta
    # --------------------------------------------------

    mfcc_delta = librosa.feature.delta(mfcc)

    mfcc_delta2 = librosa.feature.delta(
        mfcc,
        order=2
    )

    for i in range(13):

        features[f"mfcc_delta_{i+1}_mean"] = np.mean(
            mfcc_delta[i]
        )

        features[f"mfcc_delta_{i+1}_std"] = np.std(
            mfcc_delta[i]
        )

        features[f"mfcc_delta2_{i+1}_mean"] = np.mean(
            mfcc_delta2[i]
        )

        features[f"mfcc_delta2_{i+1}_std"] = np.std(
            mfcc_delta2[i]
        )

    # --------------------------------------------------
    # Speech / silence segmentation
    # --------------------------------------------------

    intervals = librosa.effects.split(
        y,
        top_db=35
    )

    if len(intervals) > 0:

        speech_samples = np.sum(
            intervals[:, 1] - intervals[:, 0]
        )

        speech_duration = speech_samples / sr

        silence_duration = max(
            duration - speech_duration,
            0
        )

        features["speech_duration"] = speech_duration
        features["silence_duration"] = silence_duration

        features["speech_ratio"] = (
            speech_duration / duration
            if duration > 0 else 0
        )

        features["silence_ratio"] = (
            silence_duration / duration
            if duration > 0 else 0
        )

        features["speech_segments"] = len(intervals)

        # --------------------------------------------------
        # Pause statistics
        # --------------------------------------------------

        if len(intervals) > 1:

            pauses = (
                intervals[1:, 0]
                - intervals[:-1, 1]
            ) / sr

            features["pause_count"] = len(pauses)

            features["pause_mean"] = np.mean(pauses)

            features["pause_std"] = np.std(pauses)

            features["pause_min"] = np.min(pauses)

            features["pause_max"] = np.max(pauses)

        else:

            features["pause_count"] = 0
            features["pause_mean"] = 0
            features["pause_std"] = 0
            features["pause_min"] = 0
            features["pause_max"] = 0

    else:

        features["speech_duration"] = 0
        features["silence_duration"] = duration
        features["speech_ratio"] = 0
        features["silence_ratio"] = 1
        features["speech_segments"] = 0
        features["pause_count"] = 0
        features["pause_mean"] = 0
        features["pause_std"] = 0
        features["pause_min"] = 0
        features["pause_max"] = 0

    # --------------------------------------------------
    # Approximate speaking rate
    # --------------------------------------------------

    features["speech_density"] = (
        duration
    )

    return features
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout,
    BatchNormalization
)


def build_multimodal_mlp(input_dim):

    model = Sequential([

        Input(shape=(input_dim,)),

        Dense(256, activation="relu"),
        BatchNormalization(),
        Dropout(0.30),

        Dense(128, activation="relu"),
        BatchNormalization(),
        Dropout(0.25),

        Dense(64, activation="relu"),
        Dropout(0.20),

        Dense(32, activation="relu"),

        Dense(1, activation="linear")
    ])

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=0.0005
        ),
        loss="mse"
    )

    return model
```

```python
from sklearn.model_selection import KFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
from scipy.stats import pearsonr

kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

multi_oof = np.zeros(len(y_multi))

fold_rmse = []
fold_pearson = []

for fold, (train_idx, val_idx) in enumerate(
    kf.split(X_multi),
    start=1
):

    print(f"\n========== FOLD {fold} ==========")

    X_train = X_multi.iloc[train_idx]
    X_val = X_multi.iloc[val_idx]

    y_train = y_multi.iloc[train_idx].values
    y_val = y_multi.iloc[val_idx].values

    # ---------------------------------------
    # Scaling
    # ---------------------------------------

    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(X_train)
    X_val_scaled = scaler.transform(X_val)

    # ---------------------------------------
    # Model
    # ---------------------------------------

    model = build_multimodal_mlp(
        X_train_scaled.shape[1]
    )

    early_stopping = EarlyStopping(
        monitor="val_loss",
        patience=25,
        restore_best_weights=True
    )

    reduce_lr = ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=8,
        min_lr=1e-6
    )

    # ---------------------------------------
    # Training
    # ---------------------------------------

    history = model.fit(
        X_train_scaled,
        y_train,
        validation_data=(
            X_val_scaled,
            y_val
        ),
        epochs=250,
        batch_size=32,
        callbacks=[
            early_stopping,
            reduce_lr
        ],
        verbose=0
    )

    # ---------------------------------------
    # Prediction
    # ---------------------------------------

    pred = model.predict(
        X_val_scaled,
        verbose=0
    ).ravel()

    pred = np.clip(
        pred,
        0,
        5
    )

    multi_oof[val_idx] = pred

    # ---------------------------------------
    # Metrics
    # ---------------------------------------

    rmse = np.sqrt(
        mean_squared_error(
            y_val,
            pred
        )
    )

    pearson = pearsonr(
        y_val,
        pred
    )[0]

    fold_rmse.append(rmse)
    fold_pearson.append(pearson)

    print(f"RMSE    : {rmse:.4f}")
    print(f"Pearson : {pearson:.4f}")
    print(f"Epochs  : {len(history.history['loss'])}")
```

# 1. Approach
The objective of this project is to build an end-to-end multimodal regression pipeline to predict spoken language proficiency and grammar scores from audio recordings. To achieve high predictive accuracy, the model combines three complementary data modalities:

Linguistic Features: Captures vocabulary richness, syntactic complexity, Part-of-Speech (POS) distributions, and fluency markers (fillers and repetitions).

Acoustic Features: Measures prosody, voice energy, spectral properties, and speech-to-silence segmentation (pauses and pacing).

Deep Semantic Embeddings: Utilizes transformer-based sentence embeddings to capture deep contextual semantic meaning from the transcripts.

# 2. Preprocessing Steps
Dataset Extraction: Unzipped and structured the dataset containing training metadata (train.csv, 769 samples), test metadata (test.csv, 216 samples), and respective audio directories.

Audio Standardization: Loaded all audio files using librosa at a standardized sampling rate of 16 kHz in mono.

Speech-to-Text Transcription: Converted raw audio waveforms into written text transcripts using OpenAI's Whisper (base model).

NLP & Linguistic Feature Extraction: Used spaCy (en_core_web_sm) for tokenization, POS tagging, and syntactic analysis to engineer 49 linguistic variables.

Semantic Vectorization: Generated normalized dense vector representations (384 dimensions) of the transcripts using the Sentence Transformer model (all-MiniLM-L6-v2).

Acoustic Feature Extraction: Extracted ~102 acoustic descriptors via librosa, including RMS energy, spectral centroids/bandwidth, 13 Mel-Frequency Cepstral Coefficients (MFCCs) with deltas, and pause/silence statistics.

# 3. Pipeline Architecture
Feature Integration: Merged linguistic features, acoustic features, and Sentence Transformer embeddings into a unified tabular matrix using the filename key.

Validation Strategy: Implemented 5-Fold Cross-Validation (KFold with shuffling and random_state=42) to robustly train and validate the model.

Data Scaling: Applied StandardScaler independently across training and validation folds to normalize input features.

Neural Network Architecture (Multilayer Perceptron):

Input Layer: Dimension matching the combined multimodal feature space.

Hidden Layers: Dense layer (256 units, ReLU activation), Batch Normalization, Dropout (0.3), Dense layer (128 units, ReLU) , Batch Normalization, Dropout (0.3)

Output Layer: Single dense unit (linear activation) with predictions clipped between 0 and 5.

Optimization & Callbacks: Trained using Adam optimizer with EarlyStopping (patience=25) and ReduceLROnPlateau (factor=0.5, patience=8).
